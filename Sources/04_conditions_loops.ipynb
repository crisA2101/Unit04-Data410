**Loops**

While

```python
i = 0
while i<4:
    print(i)
    i += 1
```

```python
i = 0
while i<10:
    print(2*i+3)
    i += 3
```

For

```python
for i in range(6):
    print (i)
```

```python
cars = ['Tesla', 'Mazda', 'Ford', 'Toyota', 'Benz']

for i in cars:
    print (i)
```

```python
all_letters = []
course = 'Business Context in Data Analytics'

for (letter in course):
    all_letters.append(letter)
    
print (all_letters);
    
```
