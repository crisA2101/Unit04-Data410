```python
import csv
```

```python
makes = ['Toyota', 'Ford', 'BMW']
models = ['RAV4', 'F150', 'X6']
max_speed = ['220', '200', '260']
powers = ['220', '200', '260']
```

```python
cars_list = list(zip(makes, models, max_speed, powers))
```

```python
with open('../Datasets/Cars.csv', 'w', newline ='') as f:
    writer = csv.writer(f)
    writer.writerow(['Make', 'Model', 'MaxSpeed', 'Power'])
    writer.writerows(cars_list)
```
